## How do Drink Preferences Vary in Everyday Life?

Gavin Kelly

### Background

Alcohol Use Disorder is a significant burden to society, and patients often struggle to curtail use outside of clinical settings. Using Ecological Momentary Assessment (EMA) to understand daily patterns of drinking allows us to understand the social, emotional, and physical context in which drinking behavior occurs. However, EMA alcohol use studies typically reduce drinking behavior to simple quantity measures. This approach overlooks the nuanced decision making process behind _what_ people choose to drink and _when_. To address this gap, we explored preference patterns between types of alcoholic and non-alcoholic drinks in real-world contexts.

Intuition and anecdotes suggest preferences vary (e.g. wine at dinner vs beer at a bar), but we lack systematic data on these patterns. Data on drink preferences is difficult to collect; they aren’t completely captured by drinking behavior. Choices for drinks are constrained by context and may reflect availability rather than actual preferences. Understanding preference patterns could reveal how context shapes decisions to drink beyond typical “drink/don’t drink” EMA measures. 

We used an experimental choice task to comprehensively measure participants’ real-time preferences. During each EMA assessment, 250 participants completed 160 choices between alcoholic and non-alcoholic drinks, resulting in 951,028 total choices. Our unique dataset combines real-time choices with rich contextual data. This allows for unprecedented examination of drink-specific preferences in diverse natural contexts. Our use of hypothetical choices for alcohol allows us to examine participants' preferences for alcohol independently from its availability. Furthermore, we present participants with diverse choices for alcoholic and non-alcoholic drinks across a range of baseline preference levels. This allows us to determine which types of drinks participants prefer, and which contexts may lead them to reverse their baseline preferences.

We present extensive descriptive statistics, data visualizations, and statistical analyses to explore these patterns in the data. Our data may indicate whether different drinks serve different functions (e.g. social lubrication, stress relief, celebration), which is unexplored in the literature. We can also determine which non-alcoholic alternatives may curtail drinking and when. This benchmark information indicates future research directions on the patterns and functions of different types of drinks. Furthermore, an understanding of real-time preferences may help adapt interventions to people and contexts.

### Variables of Interest

**subject:** Each of the 250 participants were assigned their own subject ID, 1-250.

**session:** Each time a participant completed an EMA survey, including the 160 trial choice task, this was recorded as one session.

**trial:** Each of the 160 choices participants made during a choice task was recorded as one trial, numbered 1-160 for each session.

**alc_image:** This describes the alcoholic drink option that was offered to the participants for each trial, displayed as a standardized image.

**soft_image:** This describes the non-alcoholic drink option that was offered to the participants for each trial, displayed as a standardized image. In the dataset, non-alcoholic drinks are referred to as soft drinks, even though there are options other than soft drinks.

**response:** Whether participants chose the alcoholic or non-alcoholic (soft) drink option for each trial.

**prop_alc_responses:** This describes the proportion of trials in a session that participants chose the alcoholic option.

**value_function:** For a given trial, the value_function variable describes the participants baseline preferences towards the alcoholic or non-alcoholic drink. During the baseline session, participants rated all the alcoholic and non-alcoholic drink options they would be presented with on a 1-4 scale. For a given trial, which offers a choice between a non-alcoholic and an alcoholic drink, the value_function variable subtracts the non-alcoholic rating from the alcoholic rating. A value_function value of 3 indicates a strong baseline preference towards the alcoholic drink, a value of 0 indicates baseline indifference between the drinks, and a value of -3 indicates strong preference towards the non-alcoholic drink.

**location:** For each session, participants reported the type of location they were in by selecting from a dropdown menu of 7 choices.

**social_context_overall:** For each session, participants described the people they were around by selecting from a dropdown menu of 8 choices.

**time:** For each session, the time of day that participants completed the survey was recorded.

**day_of_week:** For each session, the day of the week that participants completed the survey was recorded.

### Hypotheses

We hypothesize that the location, social context, and time of day participants completed the choice task will have a statistically significant effect on their choices for alcohol.

Furthermore, we hypothesize that the effects of choice context on drink preferences will be mediated by the type of drinks participants are offered.

Finally, we will conduct exploratory analyses to determine the grouping of alcoholic and non-alcoholic drinks that most accurately reflects participants' preferences in everyday life.

### Data Organization

#### Data Cleansing

The original dataset was stored in three .csv files. 

Baseline Data: All raw data collected in the baseline session of the study was included in this 250 row .csv file. This includes the baseline drink ratings by all 250 participants, as well as demographic and psychological measures that are not relevant to the current project. In all steps of data cleansing, variables were renamed according to a standard procedure, data was checked for anomalies and missing rows were dropped. All participants had valid baseline data, so no rows were dropped.

Survey Data: One .csv contained the survey data, with one row for each session completed. This file had over 6000 rows.  Variables such as location and social context were combined for convenience. After data cleansing, this file contained 5982 rows of data. 

Choice Task Data: The other .csv file contained the choice task data, without the information about the time, location, social context, etc. when each session was completed. Before data cleansing, this file contained over 1,000,000 rows of choices between alcohol and non-alcoholic drinks. After data cleansing, this file contained 956080 rows of data.

After cleansing all three .csv files, I merged the files so that the data for each participant was combined, and everything was stored in one .csv file.

Data Processing .Rmd scripts are included in the Supplementary Materials.

#### Data Architecture

The data have a double-nested structure, with EMA sessions nested in participants, and choice task trials nested within each session. Each participant completed an average of 23.91 sessions. The average number of trials completed per session was 159.12, meaning participants completed all 160 trials in almost every session.

After removing irrelevant variables, the data for each participant includes measures from both baseline and EMA assessments. This means that certain observations are repeated across sessions and trials. We standardized all measures and converted relevant variables to factors to create our final tidy-compliant dataset, which can be referenced in the Supplemental Materials section.

## Data Analysis

### Technique #1: Linear Mixed-Effects Models

Before examining specific drink types, we will investigate whether participants' choices for alcohol differ across different contexts. The contexts we will be testing are location, social context, time of day, and day of week. Because the data is nested, we can't run a simple Anova to test whether choices differ by location, for example. Since the data for each session are not independent, but nested within each participant, we will run Linear Mixed-Effects Models to test each contextual variable.

We used a random intercept model to control for participants' baseline differences in preferences for alcohol. Our dataset could not support random slopes as well, because including a random slope for every location type for every participant requires too much computational power. However, our hypothesis assumes trends across participants in different locations are uniform, so using random intercepts is appropriate.

We only display the output of our first model, in which we use location to predict choices for alcohol, below. We use the same approach for all four models, so we will summarize each in turn. Our data is accessible in the Supplemental Materials section.

#### Location Mixed-Effects Model

```r
ema_lead <- read.csv(file="Supplementary_Materials/ema_lead.csv") # Load data on the session level
suppressPackageStartupMessages({
library(lmerTest)
library(lme4)
    })
ema_lead$location <- relevel(factor(ema_lead$location), ref = "home")
model_location <- lmer(prop_alc_responses ~ location + (1 | subject), data = ema_lead)
summary(model_location)
```

#### Social Context Mixed-Effects Model

```r
# Make alone the baseline, as this is the social context with the most observations
ema_lead$social_context_overall <- relevel(factor(ema_lead$social_context_overall), ref = "alone")
model_social <- lmer(prop_alc_responses ~ social_context_overall + (1 | subject), data = ema_lead)
anova(model_social)
```

#### Time of Day Mixed-Effects Model

```r
model_time <- lmer(prop_alc_responses ~ time_of_day + (1 | subject), data = ema_lead)
anova(model_time)
```

#### Day of Week Mixed-Effects Model

```r
model_day <- lmer(prop_alc_responses ~ day_of_week + (1 | subject), data = ema_lead)
anova(model_day)
```

All four contexts significantly predict participants' proportion of choices for alcohol when controlling for participants' baseline preferences. Our hypothesis was confirmed; context predicts alcohol preferences with statistical significance across all participants in our sample.

Now we will investigate whether preferences for a certain type of drink is behind these contextual preferences. Our Mixed-Effects Model showed us that preferences are higher in bar/club and restaurant/cafe settings when compared to home. Intuition tells us that increased preferences for liquor might cause increased preferences for alcohol in a bar/club, and people in a restaurant/cafe might only have increased preferences for wine. We will perform statistical tests to investigate whether these intuitions hold, and preferences for a certain drink type are significantly associated with preferences for alcohol in different contexts.

We first present descriptive data visualizations to cultivate a better understanding of the data structure, with a specific focus on the alcoholic drink types, and to set the stage for our statistical analyses.

### Visualizing Alcoholic Drink Types

<img src="Figures/n_obs_plot.jpeg" width="650">

Here, we plot the total number of trials that included a given type of alcoholic drink by the value function. These counts are directly related to the size of the drink groups. Many of the alcoholic drink options we presented to participants were classified as liquor, and because this was the most common drink type, it appeared in more trials. 

The counts are also related to the baseline ratings of participants. Observe how beer has a right-skewed distribution, while wine has a left-skewed distribution. This is because baseline ratings of beer were lower than wine. Therefore, it was more likely that participants were offered a choice where they preferred a non-alcoholic drink over beer (compared to seltzer), and vice versa.

Importantly, for a given 160-trial choice task, we ensured that participants had no baseline bias towards alcoholic versus non-alcoholic drinks. This means that, for a given choice task session, the number of choices for which they preferred alcohol (at baseline) was equal to the number of choices they preferred the non-alcoholic option. This took the weights into account, so the value function values for one session would sum to zero. This also means that there are the same number of observations to the left of 0 as to the right of 0 in the above plot.
