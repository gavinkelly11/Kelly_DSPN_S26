## How do Drink Preferences Vary in Everyday Life?

Gavin Kelly

### Background

Alcohol Use Disorder is a significant burden to society, and patients often struggle to curtail use outside of clinical settings. Using Ecological Momentary Assessment (EMA) to understand daily patterns of drinking allows us to understand the social, emotional, and physical context in which drinking behavior occurs. However, EMA alcohol use studies typically reduce drinking behavior to simple quantity measures. This approach overlooks the nuanced decision making process behind _what_ people choose to drink and _when_. To address this gap, we explored preference patterns between types of alcoholic and non-alcoholic drinks in real-world contexts.

Intuition and anecdotes suggest preferences vary (e.g. wine at dinner vs beer at a bar), but we lack systematic data on these patterns. Data on drink preferences is difficult to collect; they aren’t completely captured by drinking behavior. Choices for drinks are constrained by context and may reflect availability rather than actual preferences. Understanding preference patterns could reveal how context shapes decisions to drink beyond typical “drink/don’t drink” EMA measures. 

We used an experimental choice task to comprehensively measure participants’ real-time preferences. During each EMA assessment, 250 participants completed 160 choices between alcoholic and non-alcoholic drinks, resulting in 951,028 total choices. Our unique dataset combines real-time choices with rich contextual data. This allows for unprecedented examination of drink-specific preferences in diverse natural contexts. Our use of hypothetical choices for alcohol allows us to examine participants' preferences for alcohol independently from its availability. Furthermore, we present participants with diverse choices for alcoholic and non-alcoholic drinks across a range of baseline preference levels. This allows us to determine which types of drinks participants prefer, and which contexts may lead them to reverse their baseline preferences.

We present extensive descriptive statistics, data visualizations, and statistical analyses to explore these patterns in the data. Our data may indicate whether different drinks serve different functions (e.g. social lubrication, stress relief, celebration), which is unexplored in the literature. We can also determine which non-alcoholic alternatives may curtail drinking and when. This benchmark information indicates future research directions on the patterns and functions of different types of drinks. Furthermore, an understanding of real-time preferences may help adapt interventions to people and contexts.

### Variables of Interest

**subject:** Each of the 250 participants were assigned their own subject ID, 1-250.

**session:** Each time a participant completed an EMA survey, including the 160 trial choice task, this was recorded as one session.

**trial:** Each of the 160 choices participants made during a choice task was recorded as one trial, numbered 1-160 for each session.

**alc_image:** This describes the alcoholic drink option that was offered to the participants for each trial, displayed as a standardized image.

**soft_image:** This describes the non-alcoholic drink option that was offered to the participants for each trial, displayed as a standardized image. In the dataset, non-alcoholic drinks are referred to as soft drinks, even though there are options other than soft drinks.

**response:** Whether participants chose the alcoholic or non-alcoholic (soft) drink option for each trial.

**prop_alc_responses:** This describes the proportion of trials in a session that participants chose the alcoholic option.

**value_function:** For a given trial, the value_function variable describes the participants' baseline preferences towards the alcoholic or non-alcoholic drink. During the baseline session, participants rated all the alcoholic and non-alcoholic drink options included in the choice task on a 1-4 scale. For a given trial, in which a participant chooses between a non-alcoholic and an alcoholic drink, the value_function variable subtracts the non-alcoholic baseline rating from the alcoholic baseline rating. This is an indicator of baseline preferences towards one drink or the other as a way to measure bias introduced by everyday life. A value_function value of 3 indicates a strong baseline preference towards the alcoholic drink, a value of 0 indicates baseline indifference between the drinks, and a value of -3 indicates strong preference towards the non-alcoholic drink.

**location:** For each session, participants reported the type of location they were in by selecting an option from a dropdown menu of 7 choices.

**social_context_overall:** For each session, participants described the people they were around by selecting an option from a dropdown menu of 8 choices.

**time:** For each session, the time of day that participants completed the survey was recorded.

**day_of_week:** For each session, the day of the week that participants completed the survey was recorded.

### Hypotheses

We hypothesize that the location, social context, time of day, and day of week that participants completed the choice task will have a statistically significant effect on their choices for alcohol.

Furthermore, we hypothesize that the type of drinks participants are offered will interact significantly with the different choice contexts when predicting choices for alcohol.

Finally, we will conduct exploratory analyses to re-group the alcoholic and non-alcoholic drinks in a way that most accurately reflects participants' preferences in everyday life.

### Data Organization

#### Data Cleansing

The original dataset was stored in three .csv files. 

Baseline Data: All raw data collected in the baseline session of the study was included in this 250 row .csv file. This includes the baseline drink ratings by all 250 participants, as well as demographic and psychological measures that are not relevant to the current project. In all steps of data cleansing, variables were renamed in a standard fashion, data was checked for anomalies, and missing rows were dropped. All participants had valid baseline data, so no rows were dropped.

Survey Data: The second .csv contained the survey data, with one row for each session completed including information about the participants' current time, location, social context, etc. This file had over 6000 rows.  "Select-multiple" variables such as location and social context were combined into one column. After data cleansing, this file contained 5982 rows of data. 

Choice Task Data: The third .csv file contained the choice task data from each session. Before data cleansing, this file contained over 1,000,000 rows of choices between alcohol and non-alcoholic drinks. After data cleansing, this file contained 956028 rows of data.

After cleansing all three .csv files, I merged the files by "subject" so that the data for each participant was combined, and all data was stored in a single .csv file.

Data Processing .Rmd scripts are included in the Supplementary Materials section.

#### Data Architecture

The data have a double-nested structure, with EMA sessions nested in participants, and choice task trials nested within each session. Participants completed an average of 23.91 sessions over the course of the study. The average number of trials completed per session was 159.12, meaning participants completed all 160 choice task trials in almost every session.

The data for each participant includes measures from both baseline and EMA assessments. This means that some information is repeated in the dataset across sessions and trials. We standardized all measures and converted relevant variables to factors to create our final tidy-compliant dataset, which can be referenced in the Supplementary Materials section.

## Data Analysis

### Technique #1: Linear Mixed-Effects Models

Before examining specific drink types, we will investigate whether participants' choices for alcohol differ across different contexts. The contexts we will be testing are location, social context, time of day, and day of week. Because the data is nested, we can't run a simple Anova to test whether choices differ by location, for example. Since the data for each session are not independent, but nested within each participant, we will run Linear Mixed-Effects Models to test each contextual variable.

We used a random intercept model to control for participants' baseline differences in preferences for alcohol. Our dataset could not support random slopes, because including a random slope for every location type for every participant requires too much computational power. However, we hypothesized we would observe consistent trends for all participants across different contexts, so  using only random intercepts is appropriate.

We only display the output of our first model, in which we use location to predict choices for alcohol, below. We use the same approach for all four models, so we will summarize the significant results for each in turn. Our data is accessible in the Supplemental Materials section.

#### Location Mixed-Effects Model

```r
ema_lead <- read.csv(file="Supplementary_Materials/ema_lead.csv") # Load data on the session level
suppressPackageStartupMessages({
library(lmerTest)
library(lme4)
    })
# Make home the baseline, as this is by far the location with the most observations
ema_lead$location <- relevel(factor(ema_lead$location), ref = "home")
model_location <- lmer(prop_alc_responses ~ location + (1 | subject), data = ema_lead)
summary(model_location)
```

We observe that participants exhibit significantly different preferences in every location relative to being at home except for outdoors and "other". Our outcome variable is the proportion of choices for alcohol participants made in the choice task for each session. Participants chose alcohol significantly less when they were at school or work (-6%). Participants chose alcohol significantly more when they were at a bar or club (+14%), restaurant or cafe (+10%), or someone else's home (+6%). 

As this study design is extremely novel, interpreting these effect sizes is very difficult. Since the choices participants made are purely hypothetical it is debatable how much clinical significance they have. However, we would argue that a ~14% increase in choices for alcohol at a bar/club, and ~10% increase at a restaurant/cafe could be enough to predict a meaningful increase in alcohol consumption if it is made readily available. 

#### Social Context Mixed-Effects Model

We will now conduct the same mixed-effects model, using social context as the predictor.

```r
# Make alone the baseline, as this is by far the social context with the most observations
ema_lead$social_context_overall <- relevel(factor(ema_lead$social_context_overall), ref = "alone")
model_social <- lmer(prop_alc_responses ~ social_context_overall + (1 | subject), data = ema_lead)
anova(model_social)
```

We observe that our overall model is significant. We observed multiple significant effects. Compared to being alone, participants exhibited significantly lower choices for alcohol when with a coworker or classmate (-5%), or with an "other" (-9%) (uncategorized) type of person. Participants chose alcohol significantly more when with a friend or roommate (+9%), a romantic partner (+4%), or multiple categories of people (+5%).

#### Time of Day Mixed-Effects Model

We will re-run our model using the hour of the day as the predictor variable, from 3:00pm to 10:00pm, which is when our data is concentrated.

```r
ema_time_filtered <- ema_lead %>%
  filter(hour >= 15 & hour <= 22) %>%
  mutate(hour_factor = factor(paste0(hour, ":00"), 
                              levels = paste0(15:22, ":00"), 
                              ordered = TRUE))
model_time <- lmer(prop_alc_responses ~ hour_factor + (1 | subject), data = ema_time_filtered)
anova(model_time)
```

Again, our overall model is significant. For this model, we ordered our variables to understand how choices for alcohol change over the course of an evening, rather than using hours as dummy coded variables. We observed multiple a significant linear increase in choices for alcohol as the evening went on. There was also a significant negative quadratic trend, indicating a peak in the later hours of the evening.

#### Day of Week Mixed-Effects Model

We will re-run our model a final time using the day of the week as our predictor variable.

```r
model_day <- lmer(prop_alc_responses ~ day_of_week + (1 | subject), data = ema_lead)
anova(model_day)
```

In this model, when compared to Friday, participants' choices for alcohol were significantly lower (3-5%) on every day except for Saturday. In other words, participants had the highest preference for alcohol on weekend-nights.

Across all four models, environmental and temporal contexts significantly predicted participants' proportion of choices for alcohol while controlling for individual baseline preferences. These findings confirm our first hypothesis and reveal a clear pattern: alcohol preference is heavily heavily tied to leisure and intimacy, and suppressed by obligation.

Temporally, alcohol choice increases linearly as the evening progresses, peaking later in the night. Also, choices for alcohol are  significantly higher on weekends (Fridays and Saturdays) compared to weekdays. Environmentally and socially, participants exhibited the lowest preferences for alcohol in professional or obligatory contexts, such as being at school/work (-6%) or with a coworker/classmate (-5%). Conversely, preferences increased in social and leisure settings compared to being at home or alone. Being at a bar/club (+14%), a restaurant/cafe (+10%), or a peer's home (+6%), as well as being accompanied by friends/roommates (+9%) or romantic partners (+4%), all significantly increased the proportion of alcohol choices.

Since these conclusions are being drawn from observational data explored using post-hoc hypotheses, they should not be taken as causal.

While these models indicate that context drives preferences for alcohol, they all alcoholic drinks as indistinguishable. We will now investigate whether preferences for certain alcoholic drink types drive these contextual preferences. Our Mixed-Effects Model showed us that preferences are higher in bar/club and restaurant/cafe settings when compared to home. Intuition tells us that the 14% increased preference for alcohol in a bar/club might be driven by a preference for liquor. Conversely, one might presume that people in a restaurant/cafe have increased preferences for wine. We will investigate whether these intuitions hold, and preferences for certain drink types are significantly associated with changing preferences for alcohol in different contexts.

We first present descriptive data visualizations to cultivate a better understanding of the data structure, with a specific focus on the alcoholic drink types, which will set the stage for our statistical analyses.

### Technique #2: Moderation

### Linear Mixed-Effects Models with Interaction Terms

Having established that environmental and social contexts significantly alter the overall probability of choosing alcohol, we next sought to understand what was driving these shifts. In real-world settings, choices to drink do not occur in a vacuum; rather, they depend on the specific beverages available. Our data has the advantage of offering participants choices removed from real-world constraints. To investigate how this affected preferences, we introduced interaction terms for the type of drink participants were offered into our models. 

An interaction term tests whether the effect of a specific context on preferences for alcohol is driven by specific types of drinks—for instance, whether being at a bar might increases preferences for liquor, while having a negligible effect on preferences for wine. Prior to interpreting these interaction models, comprehensive diagnostic checks were performed. All necessary statistical assumptions for mixed-effects modeling—including linearity, homoscedasticity, normality of residuals, and the absence of severe multicollinearity—were adequately met, and random intercepts were used to appropriately control for the non-independence of repeated within-subject observations.

```r
# load data in long format, with a column that includes choice proportions for all drink types
ema_long = read.csv(file="Supplementary_Materials/ema_long.csv")
```

#### Location * Drink Type

We first add alcoholic drink type as a moderator into our model where location predicts choices for alcohol.

```r
model_loc_drink <- lmer(prop_drink_type_responses ~ location * drink_type + (1 | subject), 
                        data = ema_long)
anova(model_loc_drink)
```

Here we see a significant interaction effect, meaning preference patterns for drink types change based on location. We can visualize the output of the model using the "emmeans" package for ease of interpretation:

<img src="Figures/model_loc_drink_plot.jpeg" width="750">

To analyze the interaction, we pay attention to how the patterns of drink type preferences change between locations. If the lines are semi-parallel, then the relative preferences are staying constant. For example, preferences for seltzer are predicted to be significantly higher than liquor in every location. 

On the other hand, crossing lines indicate changes in preference patterns. For example preferences for beer and wine are not significantly different than liquor except at home or another's home. This might indicate that beer and wine help people relax in calm environments, while liquor does not serve this purpose.

Furthermore, the only contexts in which seltzer is not significantly preferred over either beer or wine is in a Bar/Club or at a Restaurant. This indicates that these locations smooth preferences. Customers simply want to drink alcohol, and aren't picky about the drink type. This is also reflected by the higher overall level of choices for alcohol in these locations.

#### Social Context * Drink Type

We use alcoholic drink type as a moderator in our model in which social context predicts choices for alcohol.

```r
model_soc_drink <- lmer(prop_drink_type_responses ~ social_context_overall * drink_type + (1 | subject), data = ema_long)
anova(model_soc_drink)
```

We observe a significant interaction effect, so we visualize the interaction below:

<img src="Figures/model_soc_drink_plot.jpeg" width="750">

Again, across different social contexts, seltzer is consistently preferred over liquor. However, there are some interactions involving beer and wine. When participants were around multiple types of people or strangers, they preferred beer, wine, and liquor equally. This might indicate that in unfamiliar or uncomfortable situations, seltzer is significantly preferred to other types of drinks. Of note, overall preferences are not significantly different in these contexts, but seltzer might be the drink that participants identify as the best "social lubricant."

On the other hand, when participants were alone or with family, beer and wine were significantly preferred over liquor. Also, when participants were with their romantic partner, they significantly preferred wine over liquor. This might again indicate that beer and wine are preferred in relaxed contexts with familiar people when compared to liquor.

#### Time of Day * Drink Type

Before running the interaction with Time of Day, we filter out the hours for which we have too few observations, and convert the variable to a factor.

```r
suppressPackageStartupMessages ({ library(dplyr) })
ema_long_filtered <- ema_long %>%
  filter(hour >= 15 & hour <= 22) %>%
  mutate(hour_factor = factor(paste0(hour, ":00"), 
                              levels = paste0(15:22, ":00"), 
                              ordered = TRUE))
```

```r
model_time_drink <- lmer(prop_drink_type_responses ~ hour_factor * drink_type + (1 | subject), 
                        data = ema_long_filtered)
anova(model_time_drink)
```

We see that the interaction term is not significant, so we won't explore this model further.

#### Day of Week * Drink Type

```r
model_day_drink <- lmer(prop_drink_type_responses ~ day_of_week * drink_type + (1 | subject), 
                        data = ema_long)
anova(model_day_drink)
```

We see here that the interaction term is not significant, so we won't explore this model further.

In summary, we observed significant interactions between drink type and location, and drink type and social context. It appears that, if our goal is to identify the contexts in which drink types are important, location and social context are the most relevant. We identified that seltzer is significantly preferred over liquor across contexts. Also, beer and wine appear to be preferred in more comfortable or relaxed contexts, while they are comparable to liquor in more unfamiliar or uncomfortable contexts. In these contexts, seltzer may be viewed as best "social lubricant." While these patterns are interesting, they are solely based on observational data. They should be interpreted as future directions for experimental research and eventual interventions.

We are also interested in how non-alcoholic drinks play into these interactions. We will try regrouping the alcoholic drinks with the non-alcoholic drinks using Principal Components Analysis to determine whether there is a simpler way to group the drink options while still explaining the variance in the data.

### Technique #3: Principal Components Analysis

We will add in our 5 types of non-alcoholic drinks: Soda, Diet Soda, Juice, Energy Drinks, and Coffee.

We will take advantage of the covariance between choices for our 9 drink types to collapse them into components that will predict drinking patterns across different contexts. We must first center our data, and isolate the the drink type variables.

```r
suppressPackageStartupMessages ({ library(tidyr) })
ema_lead_clean <- ema_lead %>%
  drop_na(prop_beer_responses, prop_wine_responses, prop_liquor_responses, prop_seltzer_responses,
          prop_soda_responses, prop_diet_soda_responses, prop_juice_responses,
          prop_energy_responses, prop_coffee_responses)
ema_centered <- ema_lead_clean %>%
  select(subject, 
         prop_beer_responses, prop_wine_responses, prop_liquor_responses, prop_seltzer_responses,
         prop_soda_responses, prop_diet_soda_responses, prop_juice_responses,
         prop_energy_responses, prop_coffee_responses) %>% 
  group_by(subject) %>%
  mutate(across(starts_with("prop_"), ~ .x - mean(.x, na.rm = TRUE))) %>%
  ungroup() %>%
  select(-subject)
```

We can now run our Principal Component Analysis on the 9 drink types, where each variable stores the proportion of choices that a participant made for that drink type in a given session.

```r
pca_results <- prcomp(ema_centered, scale. = TRUE)
summary(pca_results)
```

As 2 Components explains ~78% of the variance, and we only see marginal gains when adding more components, we will use two components to model our data. Here is a graph to visualize the variance explained by each new component:

```r
plot(pca_results, type = "l", main = "Scree Plot")
```

We can see that after adding 2 components, the gains in variance we explain is marginal. So what are these components made up of?

```r
round(pca_results$rotation[1:9, 1:2], 2)
```

It appears that the first component explains variance based on whether a drink contains alcohol or not. The second component is the "vibe" component. Soda and diet soda don't contribute much to this component. Wine, Liquor, and juice contribute positively to this component, so if there is a positive value, participants are having higher alcohol % drinks, potentially mixing them with juice. 

Alternatively, beer, seltzer, energy drinks, and coffee contribute negatively to the second component. These drinks are more likely to be drunk during the day. They have a lower alcohol content and a higher caffeine content. Observing a negative value for the second component might indicate that participants are drinking more casually, consuming drinks that are more common earlier in the day. 

Interestingly, however, we did not observe a significant interaction with time of day in our mixed effects model. This may be because of the non-independence between choices for alcoholic and non-alcoholic drinks. When someone chooses more alcoholic drinks, they must choose less non-alcoholic drinks because of how the choice task is presented. Perhaps regrouping the drinks in this way is a better way to capture the importance of time of day on drink choices.

Let's see how this new "vibe" component, which groups high alcohol content/juice and low alcohol content/high caffeine drinks relates to the significant interactions we observed earlier with location and social context. We can do this by rerunning regressions, this time predicting our PC2_Vibe variable, the second principal component. 

```r
ema_lead_clean$PC1_Alcohol <- pca_results$x[, 1]
ema_lead_clean$PC2_Vibe <- pca_results$x[, 2]
ema_lead_clean$location <- relevel(factor(ema_lead_clean$location), ref = "home")
model_vibe_loc <- lmer(PC2_Vibe ~ location + (1 | subject), data = ema_lead_clean)
anova(model_vibe_loc)
```

Location significantly predicts our PC2_Vibe. Upon investigating the summary, we observe that using Home as the reference was accurate, as participants score close to zero on PC2_Vibe, meaning the vibe of being at home isn't pushing them towards one group of drinks or another. 

Being outdoors has a significant negative effect on the PC2_Vibe variable. This means that the vibe of being outdoors is leading participants to prefer low alcohol content/high caffeine. This feels reasonable, as participants might need more energy and awareness when they are outdoors.

Also, being at school or work shifts the vibe in the same direction. Understandably, participants might be less likely to prefer high alcohol content drinks such as liquor and wine when working, and would be more likely to prefer drinks high in caffeine. Here is a plot to visualize this relationship:

<img src="Figures/model_PC2_vibe_loc_plot.jpeg" width="750">

```r
ema_lead_clean$social_context_overall <- relevel(factor(ema_lead_clean$social_context_overall), ref = "alone")
model_vibe_soc <- lmer(PC2_Vibe ~ social_context_overall + (1 | subject), data = ema_lead_clean)
anova(model_vibe_soc)
```

We see a significant effect of social context on PC2_Vibe. Investigating the specific effects, we see that being around a coworker, classmate, or "other" (unclassified) type of person shifts the vibe towards the low alcohol content/high caffeine drink group. While this data is not the most informative, because the "other" group is a mystery, this reflects the finding that school/work leads participants to prefer the same drinks. It also prompts future research on other types of social groups that participants might be drinking lower alcohol content drinks with. Here is a figure to visualize this relationship:

<img src="Figures/model_PC2_vibe_soc_plot.jpeg" width="750">

### Technique #4: Group-Level k-fold Cross Validation

To test whether this vibe-based re-grouping of drinks is actually valid, we will perform cross validation to see whether our model is capable of explaining variance in a new test dataset.

Since our data is nested, we will split it by subject to ensure the model isn't training and testing on the same participants. That would be too easy! We will do 10-fold cross validation using the "rsample" package, which makes it easy to get training and test samples.

```r
suppressPackageStartupMessages({ 
  library(rsample)
  library(Metrics)
})

set.seed(403)
folds <- group_vfold_cv(ema_lead_clean, group = subject, v = 10)

cv.error.location = rep(0, 10)
cv.error.social = rep(0, 10)

start.time <- Sys.time()

for (i in 1:10){
  train_data <- analysis(folds$splits[[i]])
  test_data <- assessment(folds$splits[[i]])
  
  # Fit models with the singular check explicitly ignored
  fit_loc <- lmer(PC2_Vibe ~ location + (1 | subject), 
                  data = train_data,
                  control = lmerControl(check.conv.singular = "ignore"))
  
  fit_soc <- lmer(PC2_Vibe ~ social_context_overall + (1 | subject), 
                  data = train_data,
                  control = lmerControl(check.conv.singular = "ignore"))
  
  # Make predictions
  pred_loc <- predict(fit_loc, newdata = test_data, allow.new.levels = TRUE)
  pred_soc <- predict(fit_soc, newdata = test_data, allow.new.levels = TRUE)
  
  # Calculate errors
  cv.error.location[i] = rmse(test_data$PC2_Vibe, pred_loc)
  cv.error.social[i] = rmse(test_data$PC2_Vibe, pred_soc)
}

end.time <- Sys.time()
print(end.time - start.time)

print(paste("Location Error:", mean(cv.error.location)))
print(paste("Social Error:", mean(cv.error.social)))
```

It appears that our models are possess the same ability to predict drink preferences as determined by PC2_Vibe. In other words, the vibe that someone uses to choose their drink is equally predicted by their location as by their social context. However, are these models actually making significant predictions? We can compare the Root Mean Squared Error that we calculated for each model to the standard deviation of the data. What does this mean? We are asking whether knowing the location or social context of participants allows up to predict the PC2_Vibe component better than pure chance.

```r
sd(ema_lead_clean$PC2_Vibe)
```

As we can see, the error of our models was almost exactly the same as the standard deviation. This means that our models were barely more accurate than chance at predicting PC2_Vibe. In other words, knowing the location and social context of participants offers no predictive power for the type of drinks participants choose. It seems the only valuable principal component for predicting drink choices was the first one, which simple separated the alcoholic and non-alcoholic drinks. 

## Conclusions

Our exploratory objective was to determine if everyday drinking preferences naturally cluster into broader behavioral categories. Principal Component Analysis (PCA) successfully reduced nine specific drink types into two robust components: a primary component distinguishing alcoholic from non-alcoholic drinks, and a secondary component differentiating Low Alcohol Content/High Caffeine drinks (beer, seltzer, coffee, energy drinks) from High Alcohol Content/Juice drinks (wine, liquor, juice).

Our null findings highlight the critical importance of cross-validation and the danger of making causal claims from observational data alone. Before running the grouped cross validation, it was extremely tempting to begin to define the principal components we observed by how they might explain the variance in our sample across contexts. Alas, it turns out that our second component does not predict drinking "vibes" across locations and social contexts as well as we expected - in fact, barely better than chance!

Despite the null predictive findings regarding the specific type of drink chosen, our first hypothesis—that environmental and temporal contexts significantly affect the overall choice to consume alcohol—was robustly supported. The primary variance in our data (PC1) was driven entirely by the binary choice of alcohol versus non-alcohol.

Furthermore, our second hypothesis regarding contextual interactions was supported. The choice context exhibited significant interactions with the type of drinks participants were offered when predicting choices for alcohol. Still, these findings should be interpreted as indicators of future directions for experimental research, as they were extrapolated from observational data.

Ultimately, this project reveals a clear behavioral dichotomy in momentary drinking decisions. Environmental and temporal contexts are powerful drivers of the decision to drink alcohol; however, they do not strongly dictate the "vibe" of beverage chosen. Our regrouping of drinks to define the "vibes" of drinks indicate the importance of individual traits that participants carry with them across locations and social settings. The lack of predictive power of this "vibe" component indicates that it should be interpreted on an individual level, rather than as a state dictated by their immediate surroundings.
