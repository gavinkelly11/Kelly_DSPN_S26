## 3. Training and test error (3 points)

This question will explore the relationship between training and test error and the number of features included in a model. We will again use a simulated dataset.

a) Simulate dataset with features $p = 20$ and observations $n = 1,000$. Generate data for a response variable $Y$ according to the model

$Y = X\beta + \epsilon$

where $\beta$ is random with some elements that are exactly zero.

Split your simulated dataset into a training set containing $n=100$ observations and a test set containing $n=900$ observations.

````r
simulate_data <- function(n, p, n_zero) {
  mat <- matrix(rnorm(n * p), nrow = n, ncol = p)
  colnames(mat) <- paste0("X", 1:p)
  
  beta <- rnorm(p, mean = 2, sd = 1) 
  zero_indices <- sample(1:p, n_zero)
  beta[zero_indices] <- 0
  
  epsilon <- rnorm(n, mean = 0, sd = 1)
  
  y <- as.numeric(mat %*% beta) + epsilon
  
  return(list(
    data = data.frame(y = y, mat),
    betas = beta
  ))
}
````

````r
l1 <- simulate_data(n = 1000, p = 20, n_zero = 5)
sim_dat <- l1$data
betas <- l1$betas
````

````r
set.seed(23)
train_indices <- sample(1:1000, 100, replace = FALSE)

train_data <- sim_dat[train_indices, ]
test_data  <- sim_dat[-train_indices, ]
````

b) Perform best subset selection on the training set, and plot the associated training and test set MSE for the best model of each size.

````r
regfit.train <- regsubsets(y ~ ., data = train_data, nvmax = 20)
train.summary <- summary(regfit.train)
````

````r
predict.regsubsets <- function(object, newdata, id, ...) {
  form <- as.formula(object$call[[2]]) # Get the formula (y ~ .)
  mat <- model.matrix(form, newdata)    # Convert test data to a matrix
  coefi <- coef(object, id = id)        # Get coefficients for model of size 'id'
  xvars <- names(coefi)                 # Get names of active predictors
  mat[, xvars] %*% coefi                # Multiply data by coefficients
}
````

````r
# Initialize vectors
val_errors <- rep(NA, 20)
train_errors <- rep(NA, 20)

for (i in 1:20) {
    # 1. Predict on Training Data & Calculate MSE
    pred_train <- predict.regsubsets(regfit.train, newdata = train_data, id = i)
    train_errors[i] <- mean((train_data$y - pred_train)^2)
    
    # 2. Predict on Test Data & Calculate MSE
    pred_test <- predict.regsubsets(regfit.train, newdata = test_data, id = i)
    val_errors[i] <- mean((test_data$y - pred_test)^2)
}

# Combine into a data frame for plotting
results <- data.frame(
    n_features = 1:20,
    Train_MSE = train_errors,
    Test_MSE = val_errors
)
````

````r
library(tidyr)
library(ggplot2)

results_long <- results %>%
    pivot_longer(cols = c(Train_MSE, Test_MSE), names_to = "Dataset", values_to = "MSE")

ggplot(results_long, aes(x = n_features, y = MSE, color = Dataset)) +
    geom_line(linewidth = 1) +
    geom_point() +
    labs(title = "Train vs Test MSE by Model Size",
         x = "Number of Features",
         y = "Mean Squared Error") +
    theme_minimal()
````
