## How do Drink Preferences Vary in Everyday Life?

Gavin Kelly

### Background

Alcohol Use Disorder is a significant burden to society, and patients often struggle to curtail use outs. Using Ecological Momentary Assessment (EMA) to understand daily patterns of drinking allows us to understand the social, emotional, and physical context in which drinking behavior occurs (). However, EMA alcohol use studies typically reduce drinking behavior to simple quantity measures. This approach overlooks the nuanced decision making process behind _what_ people choose to drink and _when_. To address this gap, we explored preference patterns between types of alcoholic and non-alcoholic drinks in real-world contexts.

Intuition and anecdotes suggest preferences vary (e.g. wine at dinner vs beer at a bar), but we lack systematic data on these patterns. Data on drink preferences is difficult to collect; they aren’t completely captured by drinking behavior. Choices for drinks are constrained by context and may reflect availability rather than actual preferences. Understanding preference patterns could reveal how context shapes decisions to drink beyond typical “drink/don’t drink” EMA measures. 

We used an experimental choice task to comprehensively measure participants’ real-time preferences. During each EMA assessment, 250 participants completed 160 choices between alcoholic and non-alcoholic drinks, resulting in 951,028 total choices. Our unique dataset combines real-time choices with rich contextual data. This allows for unprecedented examination of drink-specific preferences in diverse natural contexts. Our use of hypothetical choices for alcohol allows us to examine participants' preferences for alcohol independently from its availability. Furthermore, we present participants with diverse choices for alcoholic and non-alcoholic drinks across a range of baseline preference levels. This allows us to determine which types of drinks participants prefer, and which contexts may lead them to reverse their baseline preferences.

We present extensive descriptive statistics, data visualizations, and statistical analyses to explore these patterns in the data. Our data may indicate whether different drinks serve different functions (e.g. social lubrication, stress relief, celebration), which is unexplored in the literature. We can also determine which non-alcoholic alternatives may curtail drinking and when. This benchmark information indicates future research directions on the patterns and functions of different types of drinks. Furthermore, an understanding of real-time preferences may help adapt interventions to people and contexts.

### Variables of Interest

**subject:** Each of the 250 participants were assigned their own subject ID, 1-250.

**session:** Each time a participant completed an EMA survey, including the 160 trial choice task, this was recorded as one session.

**trial:** Each of the 160 choices participants made during a choice task was recorded as one trial, numbered 1-160 for each session.

**alc_image:** This describes the alcoholic drink option that was offered to the participants for each trial, displayed as a standardized image.

**soft_image:** This describes the non-alcoholic drink option that was offered to the participants for each trial, displayed as a standardized image. In the dataset, non-alcoholic drinks are referred to as soft drinks, even though there are options other than soft drinks.

**response:** Whether participants chose the alcoholic or non-alcoholic (soft) drink option for each trial.

**prop_alc_responses:** This describes the proportion of trials in a session that participants chose the alcoholic option.

**value_function:** For a given trial, the value_function variable describes the participants baseline preferences towards the alcoholic or non-alcoholic drink. During the baseline session, participants rated all the alcoholic and non-alcoholic drink options they would be presented with on a 1-4 scale. For a given trial, which offers a choice between a non-alcoholic and an alcoholic drink, the value_function variable subtracts the non-alcoholic rating from the alcoholic rating. A value_function value of 3 indicates a strong baseline preference towards the alcoholic drink, a value of 0 indicates baseline indifference between the drinks, and a value of -3 indicates strong preference towards the non-alcoholic drink.

**location:** For each session, participants reported the type of location they were in by selecting from a dropdown menu of 7 choices.

**social_context_overall:** For each session, participants described the people they were around by selecting from a dropdown menu of 8 choices.

**time:** For each session, the time that participants completed the survey was recorded.

**day_of_week:** For each session, the day of the week that participants completed the survey was recorded.

### Hypotheses

We hypothesize that the location, social context, and time of day participants complete the choice task will have a statistically significant effect on their choices for alcohol.

Furthermore, we hypothesize that the effects of choice context on drink preferences will be mediated by the type of drinks participants are offered.

Finally, we will conduct exploratory analyses to determine the grouping of alcoholic and non-alcoholic drinks that most accurately reflects participants' preferences in everyday life.
