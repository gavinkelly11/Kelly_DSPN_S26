##How do Drink Preferences Vary in Everyday Life?

Gavin Kelly

###**Background**

AUD is a significant burden to society (). Using EMA to understand daily patterns of drinking is important (). This allows us to understand the social, emotional, and physical context in which drinking behavior occurs (). However, EMA alcohol use studies typically reduce drinking behavior to simple quantity measures. This approach overlooks the nuanced decision making process behind _what_ people choose to drink and _when_. To address this gap, we explored preference patterns between types of alcoholic and non-alcoholic drinks in real-world contexts.
Intuition and anecdotes suggest preferences vary (e.g. wine at dinner vs beer at a bar), but we lack systematic data on these patterns. Data on drink preferences is difficult to collect; they aren’t completely captured by drinking behavior. Choices for drinks are constrained by context and may reflect availability rather than actual preferences. Understanding preference patterns could reveal how context shapes decisions to drink beyond typical “drink/don’t drink” EMA measures. 
We used an experimental choice task to comprehensively measure participants’ real-time preferences. During each EMA assessment, 250 participants completed 160 choices between alcoholic and non-alcoholic drinks, resulting in 951,028 total choices. Our unique dataset combines real-time choices with rich contextual data. This allows for unprecedented examination of drink-specific preferences in diverse natural contexts.
